```python
import pandas as pd
import numpy as np

# Load diabetes diagnostic dataset
df = pd.read_csv("diabetes_prediction_dataset.csv")
df.head()
```

```python
# Inspect tail records
df.tail()
```

```python
# Dimensionality and column index
print("Dataset Shape:", df.shape)
print("Feature List:", df.columns.tolist())
```

```python
# Verify feature schema data types
df.dtypes
```

```python
# Structural summary and memory usage
df.info(memory_usage="deep")
```

```python
# Summary statistics for numerical attributes
df.describe().T
```

```python
# Age distribution metrics
print(f"Mean Age: {df['age'].mean():.2f}")
print(f"Median Age: {df['age'].median():.2f}")
print(f"Mode Age: {df['age'].mode()[0]}")
print(f"Variance: {df['age'].var():.2f}")
print(f"Standard Deviation: {df['age'].std():.2f}")
```

```python
# Age boundary and sample count
print(f"Min Age: {df['age'].min()} | Max Age: {df['age'].max()} | Total: {df['age'].count()}")
```

```python
# Smoking history categorical frequency
print("Categories:", df["smoking_history"].unique())
print("Distinct count:", df["smoking_history"].nunique())
df["smoking_history"].value_counts()
```

```python
# Null value validation across all columns
df.isna().sum()
```

```python
# Compute null value percentage per column
(df.isna().mean() * 100).round(2)
```

```python
# Compare null handling strategies
df_dropped = df.dropna()
df_imputed = df.fillna(0)

print(f"Rows remaining (dropped): {len(df_dropped)}")
print(f"Rows remaining (zero-filled): {len(df_imputed)}")
```

```python
# Check and prune duplicate records
duplicate_count = df.duplicated().sum()
print(f"Detected duplicates: {duplicate_count}")

df_deduped = df.drop_duplicates()
print(f"Records after de-duplication: {df_deduped.shape[0]}")
```

```python
# Selective slicing and column projections
df[["age", "bmi"]].head()
```

```python
# Position-based row indexing
df.iloc[:5]
```

```python
# Cohort filtering: age strictly above 30
df_over_30 = df.query("age > 30")
print("Subset count:", len(df_over_30))
df_over_30.head()
```

```python
# Ascending sort by patient age
df_sorted_age = df.sort_values(by="age", ascending=True)
df_sorted_age.head()
```

```python
# Schema adjustment: rename feature columns
df_renamed = df.rename(columns={"age": "patient_age"})
df_renamed[["patient_age", "bmi"]].head()
```

```python
# Derived feature creation and removal
df_renamed["scaled_bmi"] = df_renamed["bmi"] * 0.10
print(df_renamed[["bmi", "scaled_bmi"]].head())

df_renamed.drop(columns=["scaled_bmi"], inplace=True)
```

```python
# Grouped aggregation: average BMI by gender cohort
df.groupby("gender")["bmi"].mean().reset_index()
```

```python
# DataFrame concatenation demo
batch_a = df.iloc[:5]
batch_b = df.iloc[5:10]

stacked_batch = pd.concat([batch_a, batch_b], axis=0)
stacked_batch
```

```python
# Relational merge on synthetic patient identifier
df_indexed = df.copy()
df_indexed["patient_id"] = np.arange(len(df_indexed))

left_table = df_indexed.iloc[:5]
right_table = df_indexed.iloc[:5][["patient_id", "bmi", "diabetes"]]

merged_summary = pd.merge(left_table, right_table, on="patient_id", suffixes=('_base', '_lookup'))
merged_summary
```
