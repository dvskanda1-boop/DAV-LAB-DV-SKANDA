```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
import numpy as np

df = pd.read_csv("Salary_dataset.csv")

print("Dataset:")
print(df)

X_simple = df[["YearsExperience"]]
y = df["Salary"]

X_train_s, X_test_s, y_train, y_test = train_test_split(X_simple, y, test_size=0.2, random_state=42)

simple_model = LinearRegression()
simple_model.fit(X_train_s, y_train)

simple_predictions = simple_model.predict(X_test_s)

print("\n--- Simple Linear Regression ---")
print("Intercept:", simple_model.intercept_)
print("Coefficient:", simple_model.coef_[0])
print("R2 Score:", r2_score(y_test, simple_predictions))
print("RMSE:", np.sqrt(mean_squared_error(y_test, simple_predictions)))

new_employee = pd.DataFrame({"YearsExperience": [6]})

print("\nPredicted salary using Simple Regression:", simple_model.predict(new_employee)[0])

plt.scatter(df["YearsExperience"], df["Salary"], color="blue", label="Actual Salary")

plt.plot(
    df["YearsExperience"],
    simple_model.predict(X_simple),
    color="red",
    label="Regression Line"
)

plt.xlabel("Years Experience")
plt.ylabel("Salary")
plt.title("Simple Linear Regression")
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
import numpy as np

df = pd.read_csv("Student_Performance.csv")

print("Dataset:")
print(df)

X_multiple = df[["Hours Studied", "Previous Scores", "Sleep Hours"]]
y = df["Performance Index"]

X_train_m, X_test_m, y_train, y_test = train_test_split(X_multiple, y, test_size=0.2, random_state=42)

multiple_model = LinearRegression()
multiple_model.fit(X_train_m, y_train)

multiple_predictions = multiple_model.predict(X_test_m)

print("\n--- Multiple Linear Regression ---")
print("Intercept:", multiple_model.intercept_)
print("Coefficients:", multiple_model.coef_)
print("R2 Score:", r2_score(y_test, multiple_predictions))
print("RMSE:", np.sqrt(
    mean_squared_error(y_test, multiple_predictions)
))

new_student_multiple = pd.DataFrame({
    "Hours Studied": [6],
    "Previous Scores": [80],
    "Sleep Hours": [8]
})

print("\nPredicted performance using Multiple Regression:", multiple_model.predict(new_student_multiple)[0])
```
